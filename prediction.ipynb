```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import numpy as  np
import pandas as pd
```

```python
### Load the trained model, scaler pickle, onehot

model=load_model('model.hf')

##Load the encoder and scaler

with open('onehot_encoder_geo.pkl','rb') as file:
    onehot_encoder_geo=pickle.load(file)

with open('label_enocder_gender.pkl','rb') as file:
    label_encoder_gender=pickle.load(file)

with open('scaler.pkl','rb') as file:
    scaler=pickle.load(file)
```

```python
## Example Input Data
input_data={
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
## One Hot Encode "Geography"

geo_encoded=onehot_encoder_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df=pd.DataFrame(geo_encoded,columns=onehot_encoder_geo.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
input_df=pd.DataFrame([input_data])
input_df
```

```python
## Combine one-hot encoded columns with input_data

input_data=pd.concat([input_df.reset_index(drop=True),geo_encoded_df],axis=1)
input_data







# ## Prepare the input dataframe
# input_df=pd.DataFrame([input_data])

# ## Encode "Gender"
# input_df['Gender']=label_encoder_gender.transform(input_df['Gender'])

# ## Combine one hot encoded Geography columns with the rest of the input
# input_df=pd.concat([input_df.drop('Geography',axis=1),geo_encoded_df],axis=1)

# ## Scale the input data
# input_scaled=scaler.transform(input_df)

# ## Predict churn
# prediction=model.predict(input_scaled)
# prediction_proba=prediction[0][0]
# prediction_proba
```

```python
##Encode categorical variables
input_df['Gender']=label_encoder_gender.transform(input_df['Gender'])
input_df
```

```python
##Concatenation one hot encoded
input_df=pd.concat([input_df.drop('Geography',axis=1),geo_encoded_df],axis=1)
```

```python
input_df
```

```python
## Scaling the data

input_scaled=scaler.transform(input_df)
input_scaled
```

```python
## Predict Churn
prediction=model.predict(input_scaled)
prediction
```

```python
prediction_proba=prediction[0][0]
prediction_proba
```

```python
if prediction_proba>0.5:
    print('The customer is likely to churn.')
else:
    print('The customer is not likely to churn.')
```
